```python
import numpy as np
```

```python
#--Numpy Arrays--
list = [ 20, 27, 9, 12, 31, 17, 4, 50, 39, 35 ]
arr = np.array(list)

#1 - multiply each item by 4
print(arr * 4)

# 2-4 - mean, median, sd
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))

# 5. Index of the smallest and largest numbers
min = np.argmin(arr)
max = np.argmax(arr)

# 6. Use those indexes to get the values
print(arr[min] , arr[max])

# 7. Get the smallest and largest values directly
print(np.min(arr) , np.max(arr))
#yes, the values match that of the last step

# 8. Boolean indexing: keep only values > 20
greater_than_20 = arr[arr >20]
print(greater_than_20)


```

```python
#--2D Numpy Arrays--

matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

# 1. Shape of the array (rows, columns)
print(matrix.shape)

# 2. Third column
print(matrix[:, 2])

# 3. Second row
print(matrix[1, :])

# 4. Multiply all elements by 2
print(matrix * 2)

# 5. Mean of each column
print(np.mean(matrix, axis = 0))

# 6. Overall mean of the array
print(np.mean(matrix))

# 7. Max value in each row and min value in each column
print(np.max(matrix, axis=1))
print(np.min(matrix, axis=0)) 

# 8. Flatten the array -- takes the 3x4 gris and lines them up into a single row
flat = matrix.flatten()
print(flat)
print(flat.shape) # gives (12,) -- bc theres only one dimension

# 9. Flattened index of the smallest and largest values
min_idx = np.argmin(matrix)
max_idx = np.argmax(matrix)
print(min_idx, max_idx) 

# 10. Convert flat indices to row-column coordinates
print(np.unravel_index(min_idx, matrix.shape))
print(np.unravel_index(max_idx, matrix.shape))
# takes a single position number from a flattened array and tells you which row and column that position belongs to in the original grid.

# 11. Boolean indexing: keep only values > 15
print(matrix[matrix > 15]) 

# 12. Sort each row -- smallest to largest, axis = 1 means it sorts across each row separately
print(np.sort(matrix, axis=1))
```
