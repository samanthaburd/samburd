```python
print(df['name'].to_list())
#the reason there are more than 50 entries is because there are 2 non states included, DC and Puerto Rico
```

```python
median_lat = df['latitude'].median()
median_long = df['longitude'].median()
print(f"Median latitude: {median_lat}")
print(f"Median longitude: {median_long}")
```

```python
abs_value_median_lat = (df['latitude'] - median_lat).abs()
print(abs_value_median_lat)
abs_value_median_long = (df['longitude'] - median_long).abs()
print(abs_value_median_long)
```

```python
df['distance from median latitude'] = abs_value_median_lat
df['distance from median longitude'] = abs_value_median_long
```

```python
print(df.shape)
#I now have 52 rows and 6 columns
```

```python
#closest to median latitude
i = df['distance from median latitude'].idxmin()
print(f"Closest to median latitude: {df.loc[i , 'name']} by {df.loc[i, 'distance from median latitude']:.2f} degrees")

#farthest from median latitude
i = df['distance from median latitude'].idxmax()
print(f"Farthest from median latitude: {df.loc[i, 'name']} by {df.loc[i, 'distance from median latitude']:.2f} degrees")

#closest to median longitude
i = df['distance from median longitude'].idxmin()
print(f"Closest to median longitude: {df.loc[i, 'name']} by {df.loc[i, 'distance from median longitude']:.2f} degrees")

#farthest from median longitude
i = df['distance from median longitude'].idxmax()
print(f"Farthest from median longitude: {df.loc[i, 'name']} by {df.loc[i, 'distance from median longitude']:.2f} degrees")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset('iris')
print(data.isna().sum()) # iris has no NaN values, so dropna doesn't remove any columns.
data = data.dropna(axis=1)
print(data.shape) 
#there are 150 rows and 5 columns
```

```python
data.head()
```

```python
data.tail() 
```

```python
print(data.columns.to_list())
#sepal_length is the length of the flower's sepal in cm
#sepal_width is the width of the flower's sepal in cm
#petal_length is the length of the petal in cm
#petal_width is the width of the petal in cm
#species is the type of iris the flower is - setosa, versicolor, or virginica
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top left: sepal length vs. sepal width, colored by species (this creates the legend)
sns.scatterplot(data=data, x='sepal_length', y='sepal_width', hue='species', ax=axes[0, 0])
axes[0, 0].set_title('Sepal Length vs. Sepal Width')
axes[0, 0].set_xlabel('Sepal Length (cm)')
axes[0, 0].set_ylabel('Sepal Width (cm)')

# Top right: petal length vs. petal width
sns.scatterplot(data=data, x='petal_length', y='petal_width', color='purple', ax=axes[0, 1])
axes[0, 1].set_title('Petal Length vs. Petal Width')
axes[0, 1].set_xlabel('Petal Length (cm)')
axes[0, 1].set_ylabel('Petal Width (cm)')

# Bottom left: histogram of petal length
sns.histplot(data=data, x='petal_length', color='green', ax=axes[1, 0])
axes[1, 0].set_title('Distribution of Petal Length')
axes[1, 0].set_xlabel('Petal Length (cm)')
axes[1, 0].set_ylabel('Number of Flowers')

# Bottom right: box plot of sepal length by species
sns.boxplot(data=data, x='species', y='sepal_length', color='orange', ax=axes[1, 1])
axes[1, 1].set_title('Sepal Length by Species')
axes[1, 1].set_xlabel('Species')
axes[1, 1].set_ylabel('Sepal Length (cm)')

plt.tight_layout()
plt.show()
```

```python
#Top left (sepal length vs sepal width): I plotted sepal length vs sepal width and colored the dots by species so I could see if the species group together. Setosa (blue) is clearly separate from the other two. It has shorter but wider sepals. Versicolor and virginica are mixed together a lot, so you can’t really tell them apart just from their sepals.

#Top right (petal length vs petal width): I used the petal columns since I already used the sepal ones. The points go up in almost a straight line, so flowers with longer petals also have wider petals. There’s also a big gap between a small group in the bottom left (petals under 2 cm long) and everything else.

#Bottom left (histogram of petal length): I picked petal length because of the gap I saw in the scatter plot. The histogram shows the same thing: a tall bar around 1–1.5 cm (around 44 flowers), then basically nothing between about 2.5 and 3 cm, and then a second spread-out group from 3 to 7 cm. The first group is probably the setosa flowers.

#Bottom right (box plot of sepal length by species): I wanted to compare sepal length between the species. Setosa has the shortest sepals (median around 5 cm), versicolor is in the middle (around 5.9 cm), and virginica is the longest (around 6.5 cm). The boxes overlap some, though. There’s also one virginica flower with a really short sepal (around 4.9 cm) that shows up as an outlier dot.

#Overall: setosa is easy to tell apart, especially by petal size, but versicolor and virginica are pretty similar to each other.
```
