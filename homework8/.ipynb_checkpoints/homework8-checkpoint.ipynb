```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f"A = {A}")            
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f"ω = {w}")            
    axes[2].plot(x, make_sine_wave(x, A, w), label=f"A = {A}, ω = {w}")    

titles = ["Varying Amplitude (ω = 1)",
          "Varying Frequency (A = 1)",
          "Varying Amplitude and Frequency"]

for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=18)
    ax.set_xlabel("x (radians)", fontsize=15)
    ax.set_ylabel("y = A·sin(ωx)", fontsize=15)
    ax.legend(fontsize=12)

plt.show()
```
