```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(data.shape)
#there are 2642 rows and 2 columns
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print("Min magnitude:", magnitude.min())
print("Max magnitude:", magnitude.max())

print("Min temperature:", temperature.min())
print("Max temperature:", temperature.max())
```

```python
print("Row of min magnitude:", np.argmin(magnitude))
print("Row of max magnitude:", np.argmax(magnitude))

print("Row of min temperature:", np.argmin(temperature))
print("Row of max temperature:", np.argmax(temperature))
```

```python
#Step 4: The minimum magnitude is 2.24, which is the star in row 1988. The maximum magnitude is 12.49, which is the star in row 264. Magnitude is backwards, so row 1988 is the brightest star and row 264 is the dimmest.

#Step 5: The minimum temperature is about 3653 K, which is the star in row 1923, the coldest star. The maximum temperature is about 10113 K, which is the star in row 2613, the hottest star.

#Step 6: The brightest star is the one in row 1988, with magnitude 2.24, because it has the smallest magnitude. The dimmest star is the one in row 264, with magnitude 12.49, because it has the largest magnitude.
```

```python
# 1. Log versions of the data
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

# Row numbers of the special stars (from your argmin/argmax)
brightest = np.argmin(magnitude)    # smallest magnitude
dimmest = np.argmax(magnitude)      # largest magnitude
hottest = np.argmax(temperature)
coldest = np.argmin(temperature)

# 2. Single figure
plt.figure(figsize=(8, 6))

# 3. Scatter all stars
plt.scatter(log_temperature, log_magnitude, s=5, label="Stars")

# 5. Highlight the special stars (on the log scale)
plt.scatter(log_temperature[brightest], log_magnitude[brightest],
            color="black", marker="*", s=300, label="Brightest")
plt.scatter(log_temperature[dimmest], log_magnitude[dimmest],
            color="purple", marker="*", s=300, label="Dimmest")
plt.scatter(log_temperature[hottest], log_magnitude[hottest],
            color="green", marker="*", s=300, label="Hottest")
plt.scatter(log_temperature[coldest], log_magnitude[coldest],
            color="orange", marker="*", s=300, label="Coldest")

# 4. Invert both axes
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

# 6–9. Legend, title, labels
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")

plt.show()
```
